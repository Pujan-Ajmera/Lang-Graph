```python
import langgraph
from langgraph.graph import StateGraph,START, END
from typing import TypedDict

from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate

import dotenv


```

```python
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm = ChatGroq(
    model_name="llama-3.3-70b-versatile",
    temperature=0.7
)
prompt = PromptTemplate(
    input_variables=["text"],
    template="You are given a text and you are suppose to give me the summary of that text as given in the {text}"
)
```

```python
from typing import TypedDict

class MyClass(TypedDict):
    user_input: str
    output_text: str
```

```python
graph = StateGraph(MyClass)

def user_input(state: MyClass) -> MyClass:
    state["user_input"] = input("Enter some text: ")
    return state


def process_input(state: MyClass)-> MyClass:
    chain = prompt | llm 

    response = chain.invoke({"text":state["user_input"]})
    state["output_text"] = response.content
    return state
```

```python
graph.add_node('user_input', user_input)
graph.add_node('process_input', process_input)


graph.add_edge(START,'user_input')
graph.add_edge('user_input','process_input')

graph.add_edge('process_input',END)


workflow = graph.compile()
```

```python
initial_state = {'user_input':''}
out = workflow.invoke(initial_state)
print(out)
```
